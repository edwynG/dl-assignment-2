```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
from numpy import asarray
```

```python
# Define 4 by 4 original patch
orig_4_4 = np.array([[1, 3, 5,3 ], [6,2,0,8], [4,6,1,4], [2,8,0,3]])
print(orig_4_4)
```

```python
def downsample(x_in):
  x_out = np.zeros(( int(np.ceil(x_in.shape[0]/2)), int(np.ceil(x_in.shape[1]/2)) ))
  # TODO -- write the downsampling routine
  # Replace this line
  for i in range(x_out.shape[0]):
    for j in range(x_out.shape[1]):
      x_out[i, j] = x_in[2 * i, 2 * j]


  return x_out
```

```python
print("Original:")
print(orig_4_4)
print("Downsampled:")
print(downsample(orig_4_4))
```

```python
# Now let's try max-pooling
def maxpool(x_in):
  x_out = np.zeros(( int(np.floor(x_in.shape[0]/2)), int(np.floor(x_in.shape[1]/2)) ))
  # TODO -- write the maxpool routine
  # Replace this line

  # Buscamos el número más grande en cada bloque de 2x2
  for i in range(x_out.shape[0]):
    for j in range(x_out.shape[1]):
      x_out[i, j] = np.max(x_in[2*i : 2*i + 2, 2*j : 2*j + 2])


  return x_out
```

```python
# Finally, let's try mean pooling
def meanpool(x_in):
  x_out = np.zeros(( int(np.floor(x_in.shape[0]/2)), int(np.floor(x_in.shape[1]/2)) ))
  # TODO -- write the meanpool routine
  # Replace this line
  
  # Calculamos la media de los valores en cada bloque de 2x2
  for i in range(x_out.shape[0]):
    for j in range(x_out.shape[1]):
      x_out[i, j] = np.mean(x_in[2*i : 2*i + 2, 2*j : 2*j + 2])

  return x_out
```

```python
# Define 2 by 2 original patch
orig_2_2 = np.array([[6, 8], [8,4]])
print(orig_2_2)
```

```python
# Let's first use the duplication method
def duplicate(x_in):
  x_out = np.zeros(( x_in.shape[0]*2, x_in.shape[1]*2 ))
  # TODO -- write the duplication routine
  # Replace this line
  
  # Cada valor de entrada se copia cuatro veces en el bloque de 2x2
  for i in range(x_in.shape[0]):
    for j in range(x_in.shape[1]):
      x_out[2*i : 2*i + 2, 2*j : 2*j + 2] = x_in[i, j]


  return x_out
```

```python
print("Original:")
print(orig_2_2)
print("Duplicated:")
print(duplicate(orig_2_2))
```

```python
# Now let's try max pooling back up
# The input x_high_res is the original high res image, from which you can deduce the position of the maximum index
def max_unpool(x_in, x_high_res):
  x_out = np.zeros(( x_in.shape[0]*2, x_in.shape[1]*2 ))
  # TODO -- write the unpooling routine
  # Replace this line
  
  # Colocamos el valor en la posición exacta donde estaba el máximo original
  for i in range(x_in.shape[0]):
    for j in range(x_in.shape[1]):
      bloque = x_high_res[2*i : 2*i + 2, 2*j : 2*j + 2]
      pos_max = np.unravel_index(np.argmax(bloque), bloque.shape)
      x_out[2*i + pos_max[0], 2*j + pos_max[1]] = x_in[i, j]

  return x_out
```

Finally, we'll try upsampling using bilinear interpolation.  We'll treat the positions off the image as zeros by padding the original image and round fractional values upwards using np.ceil()

```python
def bilinear(x_in):
  x_out = np.zeros(( x_in.shape[0]*2, x_in.shape[1]*2 ))
  x_in_pad = np.zeros((x_in.shape[0]+1, x_in.shape[1]+1))
  x_in_pad[0:x_in.shape[0],0:x_in.shape[1]] = x_in
  # TODO -- write the duplication routine
  # Replace this line
  
  for i in range(x_in.shape[0]):
    for j in range(x_in.shape[1]):
      # Punto original en la esquina superior izquierda
      x_out[2*i, 2*j] = x_in_pad[i, j]
      # Promedio horizontal con el vecino de la derecha
      x_out[2*i, 2*j + 1] = np.ceil(0.5 * (x_in_pad[i, j] + x_in_pad[i, j + 1]))
      # Promedio vertical con el vecino de abajo
      x_out[2*i + 1, 2*j] = np.ceil(0.5 * (x_in_pad[i, j] + x_in_pad[i + 1, j]))
      # Centro: promedio de los 4 vecinos
      x_out[2*i + 1, 2*j + 1] = np.ceil(0.25 * (x_in_pad[i, j] + x_in_pad[i, j + 1] + x_in_pad[i + 1, j] + x_in_pad[i + 1, j + 1]))

  return x_out
```

### Resumen de lo visto


En este cuaderno aprendimos las distintas formas que existen para achicar y agrandar imágenes dentro de una red neuronal. Para reducir el tamaño, vimos que tomar valores saltados es la opción más directa pero puede perder información clave, mientras que el promedio suaviza la imagen completa y el valor máximo se queda con los puntos más llamativos y brillantes de cada zona, lo que ayuda a que el modelo reconozca figuras aunque estén un poco movidas.

Para volver a agrandar las imágenes hacia su resolución original, comprobamos que duplicar números repite los bloques pero deja la imagen pixelada y con bordes duros. Por otro lado, recordar la posición exacta de los máximos permite colocar la información justo donde nació, y la interpolación bilineal crea transiciones mucho más suaves calculando promedios entre los vecinos cercanos. Entender estas herramientas es fundamental para diseñar redes capaces no solo de clasificar fotos, sino también de reconstruirlas o recortar objetos con precisión.
